```python
import pandas as pd
from pathlib import Path

folder = Path("/Users/sharon/Downloads/dataset 2/bbc-news")

for file in folder.glob("*.csv"):

    print(f"Processing: {file.name}")

    df = pd.read_csv(file)

    # 去掉 column 空格
    df.columns = df.columns.str.strip()

    # =========================
    # rename column
    # =========================

    rename_map = {
        "Date": "date",
        "pubDate": "date",
        "Title": "title",
        "Description": "description",
        "Link": "url",
        "link": "url"
    }

    df = df.rename(columns=rename_map)

    # =========================
    # data format
    # =========================

    if "date" in df.columns:

        df["date"] = pd.to_datetime(
            df["date"],
            errors="coerce",
            dayfirst=True
        )

        df["date"] = df["date"].dt.strftime("%Y-%m-%d")

    # =========================
    # necessary cols
    # =========================

    wanted_columns = [
        "date",
        "title",
        "description",
        "url"
    ]

   
    existing_columns = [
        col for col in wanted_columns
        if col in df.columns
    ]

    df = df[existing_columns]

    # =========================
    # save
    # =========================

    output_name = file.stem + "_formatted.csv"

    df.to_csv(
        folder / output_name,
        index=False
    )

    print(f"Saved: {output_name}")
```
